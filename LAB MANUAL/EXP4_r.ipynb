# 8. Skewness

**Aim**: To calculate the skewness of a given dataset using the moments package.


**Procedure**

Install and load the moments package.

Create a vector containing the given data.

Apply the skewness() function.

Display the skewness value.

# 10. Complete Summary Statistics

**Aim**: To calculate the mean, median, variance, standard deviation and IQR of a given dataset using a single R program.

**Procedure**

Create a vector containing the dataset.

Calculate the required statistical measures.

Display each result using cat().

Display the complete summary using summary().

```r
library(moments)

data <- c(25, 28, 30, 35, 40,
          42, 45, 48, 50, 55)

cat("Mean =", mean(data), "\n")
cat("Median =", median(data), "\n")
cat("Variance =", var(data), "\n")
cat("Standard Deviation =", sd(data), "\n")
cat("IQR =", IQR(data), "\n")
cat("Skewness =", skewness(data), "\n")
cat("Kurtosis =", kurtosis(data), "\n")

summary(data)
```
