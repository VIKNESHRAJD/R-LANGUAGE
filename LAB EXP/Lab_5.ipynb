## 1. Point Estimation of Mean

**Aim:**  
To estimate the population mean from a sample of examination marks.

**Procedure:**
1. Enter the sample marks.
2. Calculate the sample mean using `mean()`.
3. Display the estimated population mean.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
marks <- c(72,68,75,80,69,74,77,71,73,76)

cat("Sample Size =", length(marks), "\n")
cat("Total Marks =", sum(marks), "\n")

mean_value <- mean(marks)
cat("Estimated Population Mean =", mean_value, "\n")

if(mean_value > 70) {
  cat("Performance is Good\n")
} else {
  cat("Performance Needs Improvement\n")
}
```

## 2. Confidence Interval for Mean

**Aim:**  
To estimate the population mean with a 95% confidence interval.

**Procedure:**
1. Enter the sample marks.
2. Apply `t.test()` with 95% confidence level.
3. Display the mean and confidence interval.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
marks <- c(72,68,75,80,69,74,77,71,73,76)

result <- t.test(marks, conf.level=0.95)

cat("Sample Mean =", result$estimate, "\n")
cat("95 Percent Confidence Interval\n")
print(result$conf.int)
cat("t Statistic =", result$statistic, "\n")
cat("Degrees of Freedom =", result$parameter, "\n")
```

## 3. Point Estimation of Proportion

**Aim:**  
To estimate the population proportion of students scoring 70 marks or above.

**Procedure:**
1. Enter the sample marks.
2. Count students scoring 70 or above.
3. Calculate and display the estimated proportion.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
marks <- c(72,68,75,80,69,74,77,71,73,76)

success <- sum(marks >= 70)
n <- length(marks)
p_hat <- success / n

cat("Students Passing =", success, "\n")
cat("Estimated Proportion =", p_hat, "\n")
cat("Percentage =", p_hat*100, "%\n")
```

## 4. Confidence Interval for Proportion

**Aim:**  
To calculate the 95% confidence interval for the population proportion of students passing an examination.

**Procedure:**
1. Enter the number of passed students and total students.
2. Apply `prop.test()`.
3. Display the estimated proportion and confidence interval.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
passed <- 150
n <- 200

result <- prop.test(passed, n, correct=FALSE)

cat("Estimated Proportion =", result$estimate, "\n")
print(result$conf.int)
cat("p-value =", result$p.value, "\n")
```

## 6. Confidence Interval for Variance

**Aim:**  
To construct a 95% confidence interval for the population variance using the Chi-square distribution.

**Procedure:**
1. Calculate the sample variance and size.
2. Find the lower and upper confidence limits.
3. Display the 95% confidence interval.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
marks <- c(72,68,75,80,69,74,77,71,73,76)

n <- length(marks)
s2 <- var(marks)
alpha <- 0.05

lower <- (n-1)*s2/qchisq(0.975,n-1)
upper <- (n-1)*s2/qchisq(0.025,n-1)

cat("Variance =", s2, "\n")
cat("95% CI =", lower, "to", upper, "\n")
```

## 7. Variance Ratio using F-Test

**Aim:**  
To compare the variances of two independent samples using the F-test.

**Procedure:**
1. Enter the two samples.
2. Apply `var.test()`.
3. Display the variance test results.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
A <- c(70,72,75,71,73,69,74,76,72,75)
B <- c(65,68,70,74,72,67,69,71,73,75)

result <- var.test(A, B)

print(result)
cat("Variance A =", var(A), "\n")
cat("Variance B =", var(B), "\n")
```

## 10. Sales Analysis of Two Stores

**Aim:**  
To analyze the sales data of two stores using mean, variance, confidence interval and variance ratio.

**Procedure:**
1. Enter the sales data for both stores.
2. Calculate mean and variance for each store.
3. Perform confidence interval and variance ratio analysis.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
storeA <- c(120,135,128,142,139,145,150,138)
storeB <- c(118,126,124,130,128,132,125,129)

cat("Store A Mean =", mean(storeA), "\n")
cat("Store B Mean =", mean(storeB), "\n")

cat("Store A Variance =", var(storeA), "\n")
cat("Store B Variance =", var(storeB), "\n")

print(t.test(storeA))
print(var.test(storeA, storeB))
```
