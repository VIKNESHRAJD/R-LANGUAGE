## 1. Create the Population

## 4. Create a Non-Normal Population

**Aim:**  
To generate an exponential population, plot its histogram, and calculate its mean.

**Procedure:**
1. Generate 10,000 exponential observations.
2. Calculate the population mean.
3. Plot the population histogram.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
set.seed(123)

population <- rexp(10000, rate = 1)

mean(population)

hist(population,
     main = "Exponential Population",
     xlab = "Values")
```

## 5. Generate 1,000 Sample Means

**Aim:**  
To generate 1,000 sample means from an exponential population using samples of size 30.

**Procedure:**
1. Set the sample size and number of samples.
2. Generate 1,000 random samples.
3. Calculate and store each sample mean.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
set.seed(123)

n <- 30
B <- 1000

sample_means <- replicate(
  B,
  mean(sample(population,
             size = n,
             replace = TRUE))
)

head(sample_means)
length(sample_means)
```

## 6. Plot the Sampling Distribution

**Aim:**  
To plot the sampling distribution of the 1,000 sample means and compare it with the population mean.

**Procedure:**
1. Plot the sample means using a histogram.
2. Mark the population mean.
3. Compare the center of both distributions.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
hist(sample_means,
     main = "Sampling Distribution of Sample Mean",
     xlab = "Sample Means",
     breaks = 30)

abline(v = mean(population), lwd = 2)

mean(population)
mean(sample_means)
```

## 7. Empirical vs Theoretical Standard Error

**Aim:**  
To compare the empirical standard error with the theoretical standard error.

**Procedure:**
1. Calculate the population standard deviation.
2. Calculate the empirical standard error.
3. Calculate and compare the theoretical standard error.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
population_sd <- sd(population)
empirical_se <- sd(sample_means)
theoretical_se <- population_sd / sqrt(n)

empirical_se
theoretical_se
```

## 8. Effect of Sample Size

**Aim:**  
To compare sampling distributions for sample sizes 5, 10, 30, and 50.

**Procedure:**
1. Set the four sample sizes.
2. Generate 1,000 sample means for each size.
3. Plot and compare the four sampling distributions.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
set.seed(123)

sample_sizes <- c(5, 10, 30, 50)

par(mfrow = c(2, 2))

for(n in sample_sizes) {
  sample_means <- replicate(
    1000,
    mean(sample(population,
                size = n,
                replace = TRUE))
  )

  hist(sample_means,
       main = paste("n =", n),
       xlab = "Sample Mean",
       breaks = 30)
}

par(mfrow = c(1, 1))
```

## 9. Demonstrating the Central Limit Theorem

**Aim:**  
To demonstrate the Central Limit Theorem using sample sizes 5, 30, and 100.

**Procedure:**
1. Generate sample means for each sample size.
2. Plot the three sampling distributions.
3. Observe the change toward a bell-shaped distribution.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
set.seed(123)

sample_sizes <- c(5, 30, 100)

par(mfrow = c(1, 3))

for(n in sample_sizes) {
  sample_means <- replicate(
    1000,
    mean(sample(population,
                size = n,
                replace = TRUE))
  )

  hist(sample_means,
       main = paste("n =", n),
       xlab = "Sample Mean",
       breaks = 30)
}

par(mfrow = c(1, 1))
```

## 10. Verify the Central Limit Theorem Numerically

**Aim:**  
To verify the Central Limit Theorem using 5,000 samples of size 30.

**Procedure:**
1. Generate 5,000 sample means.
2. Calculate population and sampling statistics.
3. Compare the empirical and theoretical standard errors.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
set.seed(123)

n <- 30
B <- 5000

sample_means <- replicate(
  B,
  mean(sample(population,
             size = n,
             replace = TRUE))
)

population_mean <- mean(population)
population_sd <- sd(population)
sampling_mean <- mean(sample_means)
empirical_se <- sd(sample_means)
theoretical_se <- population_sd / sqrt(n)

cat("Population Mean       :", population_mean, "\n")
cat("Mean of Sample Means  :", sampling_mean, "\n")
cat("Population SD         :", population_sd, "\n")
cat("Empirical SE          :", empirical_se, "\n")
cat("Theoretical SE        :", theoretical_se, "\n")

hist(sample_means,
     main = "Sampling Distribution of Sample Mean",
     xlab = "Sample Mean",
     breaks = 40)

abline(v = population_mean, lwd = 2)
```
