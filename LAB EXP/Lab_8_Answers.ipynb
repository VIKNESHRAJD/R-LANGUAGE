## F1. Variance of Rod Diameters

**Aim:**  
To test whether the population variances of rod diameters produced by two machines are significantly different.

**Procedure:**
1. Enter the diameter observations for Machine A and Machine B.
2. Perform an F-test for equality of variances using `var.test()`.
3. Use the p-value at the 5% significance level to make the decision.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
machineA <- c(20.1, 19.8, 20.3, 20.0, 19.9, 20.2)
machineB <- c(19.7, 20.4, 19.6, 20.5, 19.8, 19.5)

# F-test for equality of variances
result <- var.test(machineA, machineB)
print(result)
```

## A1. Teaching Methods – One-Way ANOVA

**Aim:**  
To test whether three teaching methods have the same population mean examination score.

**Procedure:**
1. Enter the examination scores and create the teaching-method factor.
2. Fit a one-way ANOVA model using `aov()`.
3. Display the ANOVA summary and use the p-value to make the decision.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
marks <- c(65, 68, 70, 67, 69,
            72, 75, 74, 76, 73,
            80, 82, 79, 81, 84)

method <- factor(c(rep("A", 5),
                   rep("B", 5),
                   rep("C", 5)))

data <- data.frame(marks, method)

model <- aov(marks ~ method, data = data)
summary(model)
```

## A2. Fertilizer Treatments – One-Way ANOVA

**Aim:**  
To test whether four different fertilizers produce different mean crop yields.

**Procedure:**
1. Enter the crop yields and create the fertilizer factor.
2. Fit a one-way ANOVA model.
3. Display the ANOVA summary and use the p-value at the 5% significance level.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
yield <- c(21, 23, 20, 22, 24,
            25, 27, 26, 28, 24,
            20, 19, 22, 21, 18,
            28, 30, 29, 27, 31)

fertilizer <- factor(c(rep("A", 5),
                       rep("B", 5),
                       rep("C", 5),
                       rep("D", 5)))

data <- data.frame(yield, fertilizer)

model <- aov(yield ~ fertilizer, data = data)
summary(model)
```
