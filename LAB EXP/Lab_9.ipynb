## Experiment 1 – Student Preferred Learning Method

**Aim:**

**Procedure:**

**Program:**

```r
observed <- c(36, 24, 42, 18)
result <- chisq.test(observed)
result
```

## Experiment 3 – Teaching Method and Examination Result

**Aim:**
To test the association between teaching method and examination result.

**Procedure:**
1. Create the contingency table.
2. Apply the Chi-square test of independence.
3. Display the test result and expected frequencies.

**Program:**

```r
data <- matrix(c(45,15,50,10,20,10),nrow=3, byrow=TRUE)

rownames(data) <- c("Traditional", "Blended", "Online")
colnames(data) <- c("Passed", "Failed")

result <- chisq.test(data)
result
result$expected
```

## Experiment 6 – New Medicine and Recovery

**Aim:**
To test whether the new medicine has a higher recovery rate.

**Procedure:**
1. Create the 2×2 contingency table.
2. Perform a one-sided Fisher’s Exact Test.
3. Display the test result and p-value.

**Program:**

```r
data <- matrix(c(8,2,3,7),nrow=2, byrow=TRUE)

rownames(data) <- c("New Medicine", "Standard Treatment")
colnames(data) <- c("Recovered", "Not Recovered")

result <- fisher.test(data, alternative="greater")
result
```
