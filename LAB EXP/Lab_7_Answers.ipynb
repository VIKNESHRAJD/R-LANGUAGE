## 1. One-Sample t-Test

**Aim:**  
To test whether the population mean is equal to 12 at the 5% significance level.

**Procedure:**
1. Create the sample data.
2. Perform a two-tailed one-sample t-test with hypothesized mean 12.
3. Use the p-value to make the statistical decision.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
x <- c(12, 15, 14, 10, 13, 16, 11, 14, 15, 12)

# H0: mu = 12
# H1: mu != 12

result <- t.test(x, mu = 12)

result
```

## 2. One-Sample One-Tailed t-Test

**Aim:**  
To test whether the population mean is greater than 12 at the 5% significance level.

**Procedure:**
1. Create the sample data.
2. Perform a right-tailed one-sample t-test.
3. Use the p-value to make the statistical decision.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
x <- c(12, 15, 14, 10, 13, 16, 11, 14, 15, 12)

# H0: mu <= 12
# H1: mu > 12

result <- t.test(x,
                 mu = 12,
                 alternative = "greater")

result
```

## 3. 95% Confidence Interval for Population Mean

**Aim:**  
To construct a 95% confidence interval for the population mean using a one-sample t-test.

**Procedure:**
1. Create the sample data.
2. Perform a one-sample t-test with 95% confidence level.
3. Display the confidence interval.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
x <- c(12, 15, 14, 10, 13, 16, 11, 14, 15, 12)

result <- t.test(x,
                 mu = 12,
                 conf.level = 0.95)

# Display confidence interval
result$conf.int
```

## 4. Test Against a Specified Mean

**Aim:**  
To test whether the mean mark is significantly different from 60.

**Procedure:**
1. Create the marks of eight students.
2. Perform a two-tailed one-sample t-test with mean 60.
3. Use the p-value to make the statistical decision.

**Result:**

**Program:**

```r
marks <- c(55, 60, 62, 58, 65, 61, 59, 63)

# H0: mu = 60
# H1: mu != 60

result <- t.test(marks, mu = 60)

result
```

## 5. Independent Two-Sample t-Test

**Aim:**  
To determine whether the means of two independent groups are significantly different.

**Procedure:**
1. Create the marks for Group A and Group B.
2. Perform an independent two-sample t-test.
3. Use the p-value to determine whether the population means differ.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
group_A <- c(65, 70, 72, 68, 75, 71, 69, 74)
group_B <- c(60, 62, 65, 63, 67, 61, 64, 66)

# H0: mu_A = mu_B
# H1: mu_A != mu_B

result <- t.test(group_A, group_B)

result
```

**Aim:**

## 7. One-Tailed Two-Sample t-Test

**Aim:**  
To test whether the mean of Group A is greater than the mean of Group B.

**Procedure:**
1. Create the marks for both groups.
2. Perform a right-tailed two-sample t-test.
3. Use the p-value to make the statistical decision.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
group_A <- c(65, 70, 72, 68, 75, 71, 69, 74)
group_B <- c(60, 62, 65, 63, 67, 61, 64, 66)

# H0: mu_A <= mu_B
# H1: mu_A > mu_B

result <- t.test(group_A,
                 group_B,
                 alternative = "greater")

result
```

## 8. Paired t-Test

**Aim:**  
To determine whether the training programme significantly improved the students' marks.

**Procedure:**
1. Create the before-training and after-training marks.
2. Perform a paired t-test.
3. Use the p-value to determine whether the improvement is significant.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
before <- c(55, 60, 62, 58, 65, 61, 59, 63)
after <- c(62, 65, 68, 64, 70, 67, 65, 69)

# H0: mean difference = 0
# H1: mean difference != 0

result <- t.test(before,
                 after,
                 paired = TRUE)

result
```

## 9. Formula Interface

**Aim:**  
To create a data frame with marks and group labels and perform an independent two-sample t-test using the formula interface.

**Procedure:**
1. Create the marks and group labels.
2. Create and display the data frame.
3. Use t.test(marks ~ group, data = data) to perform the test.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
# Marks
marks <- c(65, 70, 72, 68, 75, 71, 69, 74,
           60, 62, 65, 63, 67, 61, 64, 66)

# Group labels
group <- c(rep("A", 8), rep("B", 8))

# Create data frame
data <- data.frame(
  marks = marks,
  group = group
)

# Display data
print(data)

# Two-sample t-test
result <- t.test(marks ~ group, data = data)

result
```

## 10. Extract Individual Test Results

**Aim:**  
To perform a two-sample t-test, extract individual test results, and state the statistical decision at the 5% significance level.

**Procedure:**
1. Perform the independent two-sample t-test.
2. Extract the t-statistic, degrees of freedom, p-value, confidence interval, and estimated means.
3. Use the p-value with alpha = 0.05 to make the decision.

**Result:**  
Thus, the given R program was successfully executed and the required result was obtained.

**Program:**

```r
group_A <- c(65, 70, 72, 68, 75, 71, 69, 74)
group_B <- c(60, 62, 65, 63, 67, 61, 64, 66)

# Perform t-test
result <- t.test(group_A, group_B)

# Complete result
result

# t-statistic
result$statistic

# Degrees of freedom
result$parameter

# p-value
result$p.value

# Confidence interval
result$conf.int

# Estimated means
result$estimate

# Test method
result$method

# Decision using p-value
alpha <- 0.05

if (result$p.value < alpha) {
  print("Reject H0")
} else {
  print("Do not reject H0")
}
```
